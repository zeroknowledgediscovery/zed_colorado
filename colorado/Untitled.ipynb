```python
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


def _split_pm_series(series):
    means = []
    errs = []

    for v in series:
        if pd.isna(v):
            means.append(np.nan)
            errs.append(0.0)
            continue

        s = str(v).strip()
        s = s.strip("$").strip()
        s = s.replace("\\pm", "±")
        s = s.replace("+/-", "±")
        s = re.sub(r"\s+", " ", s).strip()

        m = re.match(
            r'^([+-]?\d*\.?\d+(?:[eE][+-]?\d+)?)\s*±\s*([+-]?\d*\.?\d+(?:[eE][+-]?\d+)?)$',
            s
        )
        if m:
            means.append(float(m.group(1)))
            errs.append(float(m.group(2)))
        else:
            means.append(float(s))
            errs.append(0.0)

    return np.array(means, dtype=float), np.array(errs, dtype=float)




def plot_metric_vs_pred_week(
    df,
    metric="AUC",
    pred_week_col="pred_w",
    kind="errorbar",
    title=None,
    xlabel="Prediction week",
    ylabel=None,
    sort_x=True,
    figsize=(8, 5),
    capsize=5,
    rotation=0,
    ax=None,
    label=None,
    linewidth=1.2,
    elinewidth=2.0,
    markersize=7,
    capthick=2.0,
    max_pred_week=52,
    min_pred_week=None,
):
    if pred_week_col not in df.columns:
        if df.index.name == pred_week_col:
            df = df.reset_index()
        else:
            raise ValueError(f"Column '{pred_week_col}' not found in dataframe.")

    if metric not in df.columns:
        raise ValueError(f"Column '{metric}' not found in dataframe.")

    plot_df = df[[pred_week_col, metric]].copy()
    plot_df[f"{metric}_mean"], plot_df[f"{metric}_err"] = _split_pm_series(plot_df[metric])

    # Optional week filtering
    if min_pred_week is not None:
        plot_df = plot_df[plot_df[pred_week_col] >= min_pred_week]
    if max_pred_week is not None:
        plot_df = plot_df[plot_df[pred_week_col] <= max_pred_week]

    if plot_df.empty:
        raise ValueError("No rows remain after applying pred_week limits.")

    if sort_x:
        plot_df = plot_df.sort_values(pred_week_col).reset_index(drop=True)

    x = plot_df[pred_week_col].values
    y = plot_df[f"{metric}_mean"].values
    yerr = plot_df[f"{metric}_err"].values

    if ax is None:
        fig, ax = plt.subplots(figsize=figsize)

    if kind == "bar":
        ax.bar(
            x, y, yerr=yerr,
            capsize=capsize,
            label=label,
            alpha=0.85,
            linewidth=linewidth
        )
    elif kind == "errorbar":
        ax.errorbar(
            x, y, yerr=yerr,
            fmt='o-',
            capsize=capsize,
            label=label,
            linewidth=linewidth,
            elinewidth=elinewidth,
            markersize=markersize,
            capthick=capthick
        )
    else:
        raise ValueError("kind must be 'bar' or 'errorbar'")

    ax.set_xlabel(xlabel)
    ax.set_ylabel(ylabel if ylabel is not None else metric)
    ax.set_title(title if title is not None else f"{metric} vs {pred_week_col}")
    ax.set_xticks(x)
    ax.tick_params(axis='x', rotation=rotation)
    ax.grid(True, axis='y', alpha=0.3)

    return ax, plot_df
```

```python
#ax, used = plot_metric_vs_pred_week(ILD1, metric="AUC", kind="errorbar",label='FILD')
ax, used = plot_metric_vs_pred_week(ILD2, metric="AUC", kind="errorbar",label='FILD')
ax, used = plot_metric_vs_pred_week(NFILA, metric="AUC", kind="errorbar",ax=ax,label='NFILA')
ax, used = plot_metric_vs_pred_week(NFILD, metric="AUC", kind="errorbar",ax=ax,label='NFILD')
plt.legend()
plt.tight_layout()
ax, used = plot_metric_vs_pred_week(ILD2, metric="LR+", kind="errorbar",label='FILD')
ax, used = plot_metric_vs_pred_week(NFILA, metric="LR+", kind="errorbar",ax=ax,label='NFILA')
ax, used = plot_metric_vs_pred_week(NFILD, metric="LR+", kind="errorbar",ax=ax,label='NFILD')
plt.legend()
plt.tight_layout()
```

```python

fig, axes = plt.subplots(2,1, figsize=(5, 8), sharex=True)

# Panel 1: AUC
ax = axes[0]
ax, _ = plot_metric_vs_pred_week(ILD2,  metric="AUC", kind="errorbar", ax=ax, label="FILD")
ax, _ = plot_metric_vs_pred_week(NFILA, metric="AUC", kind="errorbar", ax=ax, label="NFILA")
ax, _ = plot_metric_vs_pred_week(NFILD, metric="AUC", kind="errorbar", ax=ax, label="NFILD")
ax.set_title("AUC vs Prediction Week")
ax.legend()

# Panel 2: LR+
ax = axes[1]
ax, _ = plot_metric_vs_pred_week(ILD2,  metric="LR+", kind="errorbar", ax=ax, label="FILD")
ax, _ = plot_metric_vs_pred_week(NFILA, metric="LR+", kind="errorbar", ax=ax, label="NFILA")
ax, _ = plot_metric_vs_pred_week(NFILD, metric="LR+", kind="errorbar", ax=ax, label="NFILD")
ax.set_title("LR+ vs Prediction Week")
ax.legend()

plt.tight_layout()
plt.savefig("auc_lrplus_two_panel.png", dpi=300, bbox_inches="tight")
```
